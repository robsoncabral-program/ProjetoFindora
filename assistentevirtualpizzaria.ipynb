```
import string
from sklearn.tree import DecisionTreeClassifier

# --- Fase 3: Machine Learning (Configuração do Modelo) ---
# Dados de treino: [Idade, Salário] -> 0 (Não compra), 1 (Compra)
X = [[20, 500], [30, 2000], [50, 1500], [25, 3000], [60, 500]]
y = [0, 1, 1, 1, 0]
modelo = DecisionTreeClassifier()
modelo.fit(X, y)

def assistente_pizzaria():
    print("--- Assistente da Pizzaria (Fase Final) ---")

    while True:
        # Fase 1 & 2: Loop e Limpeza
        entrada = input("\nTu: ").lower()
        texto = "".join([c for c in entrada if c not in string.punctuation])

        # Fase 3: Intenções
        if any(p in texto for p in ["adeus", "xau", "obrigado"]):
            print("Pizzaria: Obrigado pela preferência! Até breve.")
            break
        elif any(p in texto for p in ["ola", "bom dia", "boa tarde"]):
            print("Pizzaria: Olá! Bem-vindo à nossa Pizzaria. Em que posso ajudar?")
        elif any(p in texto for p in ["menu", "ementa", "comida"]):
            print("Pizzaria: O nosso menu:\n1. Pizza Margherita (10€)\n2. Pepperoni (12€)")
        elif any(p in texto for p in ["horario", "horas", "abertos"]):
            print("Pizzaria: Estamos abertos todos os dias das 12h às 23h.")
        elif any(p in texto for p in ["onde", "morada", "localizacao"]):
            print("Pizzaria: Estamos localizados na Rua do Código, nº 42.")
        elif "prever" in texto or "compra" in texto:
            i = int(input("Insere a tua idade: "))
            s = int(input("Insere o teu salário: "))
            res = "provavelmente fará uma compra!" if modelo.predict([[i, s]])[0] == 1 else "não fará uma compra hoje."
            print(f"Pizzaria: Com base nos nossos dados, tu {res}")
        else:
            print("Pizzaria: Desculpa, não entendi. Podes perguntar sobre o menu, horário ou morada?")

# Iniciar programa
assistente_pizzaria()
```
